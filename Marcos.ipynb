#3 - Onde estão as melhores oportunidades? Quais países da residência do empregado têm os maiores salários médios? Existe correlação entre país e modalidade de trabalho (remoto vs presencial)?

```python
def media_salario_por_pais(df):
    return df.groupby('employee_residence')['salary_in_usd'].mean().round(2).reset_index().rename(columns={'employee_residence': 'País',
              'salary_in_usd': 'Salário Médio'}).sort_values(by='Salário Médio', ascending=False).head(10)
```

```python
media_salario_por_pais(df)
```

```python
df_top10 = media_salario_por_pais(df)

df_top10.plot(
    x='País',
    y='Salário Médio',
    kind='bar',
    figsize=(10,6),
    legend=False,
    color='skyblue'
)

plt.title('Top 10 países por média salarial')
plt.ylabel('Salário Médio (USD)')
plt.xlabel('País')
plt.xticks(rotation=45)
plt.show()
```

```python
def media_salario_top10_por_modalidade(df):
    top10_paises = (df.groupby('employee_residence')['salary_in_usd'].mean().round(2).reset_index()
          .sort_values(by='salary_in_usd', ascending=False).head(10)['employee_residence'].tolist())

    df_filtrado = df[df['employee_residence'].isin(top10_paises)]

    return df_filtrado.groupby(['employee_residence', 'work_models'])['salary_in_usd']\
        .mean().round(2).reset_index().rename(columns={'employee_residence': 'País',
        'work_models': 'Modalidade','salary_in_usd': 'Salário Médio'}).sort_values(by=['País', 'Salário Médio'], ascending=[True, False])
```

```python
media_salario_top10_por_modalidade(df)
```
